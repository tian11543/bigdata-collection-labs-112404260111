```python
from bs4 import BeautifulSoup
import re

rows = soup.select(".property")   

for row in rows:
    
    title_tag = row.select_one("h3")
    title = title_tag.get_text(strip=True) if title_tag else "无"

    
    layout_tag = row.select_one(".property-content-info-text:nth-child(1)")
    layout = layout_tag.get_text(strip=True) if layout_tag else "无"

    area_tag = row.select_one(".property-content-info-text:nth-child(2)")
    area = area_tag.get_text(strip=True) if area_tag else "无"

    orientation_tag = row.select_one(".property-content-info-text:nth-child(3)")
    orientation = orientation_tag.get_text(strip=True) if orientation_tag else "无"

    floor_tag = row.select_one(".property-content-info-text:nth-child(4)")
    floor = floor_tag.get_text(strip=True) if floor_tag else "无"

    year_tag = row.select_one(".property-content-info-text:nth-child(5)")
    if year_tag:
        build_year = year_tag.get_text(strip=True).replace("年建造","")
    else:
        build_year = "无"  

   
    price_box = row.select_one(".property-price")
    if price_box:
        total_tag = price_box.select_one(".property-price-total")
        unit_tag = price_box.select_one(".property-price-average")
        total_price = total_tag.get_text(strip=True) if total_tag else "无"
        unit_price = unit_tag.get_text(strip=True) if unit_tag else "无"
    else:
        total_price = "无"
        unit_price = "无"

    item = {
        "title": title,
        "layout": layout,
        "area": area,
        "orientation": orientation,
        "floor": floor,
        "build_year": build_year,
        "total_price": total_price,
        "unit_price": unit_price
    }
    print(item)
```

```python
import time, csv
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from pathlib import Path

# ========== 1. 采集 ==========
HEADERS = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
           "AppleWebKit/537.36 (KHTML, like Gecko) Chrome/141.0.0.0 Safari/537.36"}

def parse_house(row):
    title = row.select_one("h3").get_text(strip=True)
    house = row.select_one("p.property-content-info-comm-name").get_text(strip=True)
    address = row.select_one("p.property-content-info-comm-address").get_text(strip=True)
    struct = row.select_one("p.property-content-info-attribute").get_text(strip=True)
    infos = row.select("p.property-content-info-text")
    area = infos[1].get_text(strip=True)
    floor = infos[3].get_text(strip=True)
    maketime = infos[4].get_text(strip=True) if len(infos) > 4 else ""
    price = row.select_one("p.property-price-average").get_text(strip=True)
    return [title, house, address, struct, area, floor, maketime, price]

current_url = "http://www.bspider.top/anjuke/"
visited, out_list = set(), []
page_no = 0

while True:
    if current_url in visited:
        print("链接重复，停止采集"); break
    visited.add(current_url)
    page_no += 1
    if page_no > 20:
        print("达到安全页数上限，中止"); break

    resp = requests.get(current_url, headers=HEADERS, timeout=30)
    resp.raise_for_status()
    resp.encoding = "utf-8"
    page_soup = BeautifulSoup(resp.text, "html.parser")

    if page_soup.title and "访问验证" in page_soup.title.get_text():
        print("遇到访问验证页，停止采集"); break

    rows = page_soup.select("div.property")
    for row in rows:
        out_list.append(parse_house(row))
    print(f"第{page_no}页 {current_url} 本页{len(rows)}条 累计{len(out_list)}条")

    next_tag = page_soup.select_one("a.next.next-active")
    if not next_tag:
        print("下一页按钮已禁用，到达末页，采集完成"); break
    next_href_raw = next_tag.get("href")
    print(f"原始href：{next_href_raw} → 拼接后：{urljoin(current_url, next_href_raw)}")
    current_url = urljoin(current_url, next_href_raw)
    time.sleep(1)

# ========== 2. 去重 ==========
before = len(out_list)
seen, unique_rows = set(), []
for rec in out_list:
    key = tuple(rec)
    if key not in seen:
        seen.add(key)
        unique_rows.append(rec)
print(f"去重前：{before} 条，重复：{before - len(unique_rows)} 条，去重后：{len(unique_rows)} 条")

# ========== 3. 保存 ==========
header = ["title", "house", "address", "struct", "area", "floor", "maketime", "price"]
out_dir = Path("data"); out_dir.mkdir(exist_ok=True)

with open(out_dir / "houseinfo.csv", "w", encoding="utf-8-sig", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(unique_rows)
print("已保存 data/houseinfo.csv")

with open(out_dir / "houseinfo_raw.csv", "w", encoding="utf-8-sig", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerows(out_list)
print("已保存 data/houseinfo_raw.csv")

# ========== 4. 回读核对 ==========
with open("data/houseinfo.csv", "r", encoding="utf-8-sig", newline="") as f:
    reader = csv.DictReader(f)
    print("表头：", reader.fieldnames)
    data = list(reader)

print("回读数据行数（不含表头）：", len(data))
print("第一条：", data[0])
print("中间一条：", data[len(data)//2])
print("最后一条：", data[-1])
```

```python
import csv
import pymysql

conn = pymysql.connect(
    host="mysql",
    port=3306,
    database="smartpark_lab",
    user="student",
    password="123456",
    charset="utf8mb4",
    autocommit=False
)
cur = conn.cursor()

create_sql = """
CREATE TABLE IF NOT EXISTS houseinfo (
    id INT AUTO_INCREMENT PRIMARY KEY,
    title VARCHAR(255),
    house VARCHAR(255),
    address VARCHAR(100),
    struct VARCHAR(100),
    area VARCHAR(100),
    floor VARCHAR(100),
    maketime VARCHAR(100),
    price VARCHAR(100)
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4;
"""
cur.execute(create_sql)

records = []
cols = ["title","house","address","struct","area","floor","maketime","price"]
with open("data/houseinfo.csv","r",encoding="utf-8-sig") as f:
    reader = csv.DictReader(f)
    for row in reader:
        one = (
            row["title"],
            row["house"],
            row["address"],
            row["struct"],
            row["area"],
            row["floor"],
            row["maketime"],
            row["price"]
        )
        records.append(one)

print(f"读取到待入库 {len(records)} 条数据")


insert_sql = """
INSERT INTO houseinfo (title,house,address,struct,area,floor,maketime,price)
SELECT %s,%s,%s,%s,%s,%s,%s,%s
FROM DUAL
WHERE NOT EXISTS (
    SELECT 1 FROM houseinfo
    WHERE title=%s AND house=%s AND address=%s AND struct=%s AND area=%s AND floor=%s AND maketime=%s AND price=%s
)
"""


records_check = []
for r in records:
    records_check.append(r + r)


try:
   
    cur.executemany(insert_sql, records_check)
    conn.commit()  
    print(f"成功入库，受影响行数：{cur.rowcount}")
except Exception as e:
    conn.rollback() 
    print("入库失败，回滚！错误：", e)
finally:
    cur.close()
    conn.close()
    print("游标、连接已关闭")
```
